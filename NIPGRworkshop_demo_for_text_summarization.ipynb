```python
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

article = """Chapter 2 assesses observed large-scale changes in climate system drivers, key climate indicators and principal modes of variability. Chapter 3 considers model performance and detection/attribution, and Chapter 4 covers projections for a subset of these same indicators and modes of variability. Collectively, these chapters provide the basis for later chapters, which focus upon processes and regional changes. Within Chapter 2, changes are assessed from in situ and remotely sensed data and products and from indirect evidence of longer-term changes based upon a diverse range of climate proxies. The time-evolving availability of observations and proxy information dictate the periods that can be assessed. Wherever possible, recent changes are assessed for their significance in a longer-term context, including target proxy periods, both in terms of mean state and rates of change.
"""
```

```python
testsumm = summarizer(article, max_length=130, min_length=30, do_sample=False)
testsumm[0]['summary_text']
```
